```python
import seaborn as sns
import matplotlib.pyplot as plt

data_set = sns.get_dataset_names()
```

```python
print(data_set)
```

```python
taxi = sns.load_dataset("taxis")
taxi.head()
```

```python
# plt.subplot(1,2,1)
sns.lineplot(x="passengers",y="fare",linestyle="--",color="red",markeredgecolor="tab:red",data=taxi)

# plt.subplot(1,2,2)
sns.scatterplot(x="passengers",y="tip",color="green",data=taxi)

plt.tight_layout()

ss=taxi.head()
ss
```
