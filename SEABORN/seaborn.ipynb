```python
import seaborn as sns
import matplotlib.pyplot as plt
data_set = sns.get_dataset_names()
```

```python
taxi = sns.load_dataset("taxis")
taxi.head()
```

```python
# plt.subplot(1,2,1)
sns.lineplot(x="passengers",y="fare",linestyle="--",color="red",markers=True,markeredgecolor="tab:red",data=taxi)

# plt.subplot(1,2,2)
sns.scatterplot(x="passengers",y="tip",color="green",data=taxi)
# sns.(x="passengers",y="fare",marker="X",data=taxi)

plt.tight_layout()
```

```python
import pandas as pd
data = pd.read_csv("da.csv")

plt.figure(figsize=(20,20))
sns.lineplot(x="Name",y="Marks",marker="o",markeredgecolor="tab:green",color="red",data=data)
plt.grid()
plt.xticks(fontsize=10, fontfamily="serif",fontweight="bold")  # rotation=45
plt.yticks(fontsize=10, fontweight="bold",fontfamily="serif")

plt.title(
    "Sales Analysis",
    fontsize=16,
    fontweight="bold",
    fontstyle="italic",
    fontfamily="serif",
)
plt.xlabel("Name",fontsize=16,
    fontweight="bold",
    fontstyle="italic",
    fontfamily="serif")
plt.ylabel("Marks", fontsize=16,
    fontweight="bold",
    fontstyle="italic",
    fontfamily="serif")
```

```python
plt.figure(figsize=(20,20))
# sns.set_theme(style="ticks", palette="deep")
sns.relplot(x="Name",y="Marks",data=data,height=8,aspect=2)
plt.grid()
plt.xticks(fontsize=10, fontfamily="serif",fontweight="bold")  # rotation=45
plt.yticks(fontsize=10, fontweight="bold",fontfamily="serif")

plt.title(
    "Sales Analysis",
    fontsize=16,
    fontweight="bold",
    fontstyle="italic",
    fontfamily="serif",
)
plt.xlabel("Name",fontsize=16,
    fontweight="bold",
    fontstyle="italic",
    fontfamily="serif")
plt.ylabel("Marks", fontsize=16,
    fontweight="bold",
    fontstyle="italic",
    fontfamily="serif")
sns.lineplot(x="Name",y="Marks",color="red",data=data)
plt.tight_layout()
```
