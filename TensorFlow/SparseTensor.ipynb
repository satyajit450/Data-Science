```python
import tensorflow as tf 
import numpy as np
```

```python
tf_todense = tf.sparse.SparseTensor(indices =[[0,0],[1,1],[2,2],[3,3]] ,values =[1,2,3,4] ,dense_shape=(4,4))   ## indices = [row,column] in which value are stored , values = which value are stored in the matrix , dense_shape = shape of matrix 1D,2D,3D
```

```python
tf_todense
```

```python
tf_tosparse  = tf.sparse.to_dense(tf_todense)
```
