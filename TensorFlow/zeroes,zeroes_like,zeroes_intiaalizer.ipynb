```python
import tensorflow as tf
import numpy as np
```

```python
ones = tf.ones([2,2],dtype = tf.int32)
ones
```

```python
ones_ = tf.ones([2,2,4],dtype = tf.int32)  # it direct creates ones
ones_
```

# Ones_ like

```python
ones_like_1D = tf.ones_like(ones_,dtype = tf.int32)

print(ones_like_1D)  # one_like take input to create ones matrix
```

# Ones_Intializer

```python
def ones_intializer(shape , dtype , intializer) :
    return tf.Variable(intializer(shape=shape,dtype=dtype))
```

```python
print(ones_intializer((2,2), tf.int32 , tf.ones_initializer()))
```
