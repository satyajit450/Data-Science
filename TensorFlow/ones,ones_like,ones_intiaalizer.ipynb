```python
import tensorflow as tf
import numpy as np  
## In TensorFlow, the difference between tf.ones() and tf.ones_like() is the same idea as in NumPy or PyTorch — but applied to tensors.
```

```python
ones = tf.ones((2,2),dtype=tf.int32,name = "Zzz")
ones
```

```python
def one_intializer(shape , dtype, intializer) :
    return tf.Variable(intializer(shape=shape , dtype = dtype))
```

```python
ones = one_intializer((2,2),tf.int32, tf.ones_initializer())
ones
```

```python
import tensorflow as tf

initializer = tf.keras.initializers.Ones()

# Example: Dense layer with Ones initializer
layer = tf.keras.layers.Dense(
    units=3,
    kernel_initializer=initializer,
    bias_initializer=initializer
)

# Dummy input
x = tf.constant([[1.0, 2.0]])
y = layer(x)

print("Weights:\n", layer.kernel.numpy())
print("Bias:\n", layer.bias.numpy())
```
