```python
import requests 

url = "https://www.gutenberg.org/"
response = requests.get(url)

with open("books/book.html","w",encoding="utf-8") as f :
    f.write(response.text)
```

```python
from bs4 import BeautifulSoup
import pandas as pd
data = []
with open("books/book.html","r",encoding="utf-8") as f:
    soup = BeautifulSoup(f,"html.parser")

    title = soup.select("div.library div.box_shadow div.lib.latest.no-select a")
    # print(title)
    
    for target in title:
       titles = target.get('title')
       Author = target.get('data-authors')
       data.append([titles,Author])
data

df = pd.DataFrame(data , columns =["Titles","Author"])
df.to_csv("data3.csv")
```
