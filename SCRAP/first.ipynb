```python
import requests
for i in range (1,51):
    url = f"https://books.toscrape.com/catalogue/page-{i}.html"
    response = requests.get(url)
    with open(f"htmls/page{i}.html","w",encoding="utf-8") as f :
        f.write(response.text)
    print(f"page{i} completed")
```

```python
from bs4 import BeautifulSoup

data = []

for i in range(1, 51):
    with open(f"htmls/page{i}.html", "r", encoding="utf-8") as f:
        soup = BeautifulSoup(f, "html.parser")
    
        print(f"{i} page completed")
    
    for articles in soup.find_all("article"):
        title_tag = articles.find("h3").find("a")["title"]
        price = articles.select_one("div.product_price > p.price_color").text.split("£")[1]
        star_rating = articles.select_one("p.star-rating")
        rating = star_rating["class"][1]
        data.append([title_tag,price,rating])
        
    data
```

```python
import pandas as pd 
df = pd.DataFrame(data,columns=["Title","Price","Rating"])
df.to_csv("data1.csv",encoding="utf-8")
df
```
