```python
import requests
for i in range (1,51):
    url = f"https://books.toscrape.com/catalogue/page-{i}.html"
    response = requests.get(url)
    with open(f"htmls/page{i}.html","w",encoding="utf-8") as f :
        f.write(response.text)
    print(f"page{i} completed")
```

```python
from bs4 import BeautifulSoup

data = []

# for i in range(1, 51):
with open("htmls/page1.html", "r", encoding="utf-8") as f:
    soup = BeautifulSoup(f, "html.parser")

    # print(f"{i} page completed")

for article in soup.find_all("article"):
    title_tag = article.find("h3")
    title = title_tag.text
    price = article.select_one("div.product_price > p.price_color").text.split("£")[1]
    star_rating = article.select_one("article.product_pod p.star-rating").text
    data.append([title,price,star_rating])
    
```

```python
import pandas as pd 
df = pd.DataFrame(data,columns=["Title","Price","Rating"])
df.to_csv("data1.csv",encoding="utf-8")
```
