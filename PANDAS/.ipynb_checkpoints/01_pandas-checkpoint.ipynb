```python
#17. Aggregation & GroupBy 
# Find:
#mean 
#sum
#count
#max
#min

df2 = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
        "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})  
```

```python
df2.groupby("Department")["Salary"].mean()
df2.groupby("Team")["Age"].max()
```

```python
df2.groupby("Department")["Salary"].agg(
    avg_salary = "max",
    Age = "max",
    date = "count"
)
```

```python
# 18. Group data by a categorical column and calculate:
# 1. average
# 2. total

df2.groupby("Department")["Salary"].mean()
df2.groupby("Department")["Salary"].sum()
```

```python
# 19. Sort values by:
# 1. one column
# 2. multiple columns

df2.sort_values("Salary",ascending=False)
df2.sort_values(["Salary","Age"],ascending=False)
```

```python
# 20. Create a new column based on a condition.
df2 = df2.drop_duplicates()
import numpy as np

mean_salary = df2["Salary"].mean()
max_salary = df2["Salary"].max()

df2["Mean Salary"] = np.where(
    df2["Gender"] == "M",    #if not not use pandas
    mean_salary,
    max_salary
)

df2
```

```python
# 21. Use .apply() with a custom function. 
df2["Young_Old"] = df2['Age'].apply(lambda x : "Young" if x <= 25 else "Old")
df2
```

```python
# 21. Use .map() with a custom function. 
gender = {"M" : "Male","F":"Female"}
df2["Gender"]=df2["Gender"].map(gender)
```
