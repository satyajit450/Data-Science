```python
#17. Aggregation & GroupBy 
# Find:
#mean 
#sum
#count
#max
#min

df2 = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],                                                                                                        
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [18, 30, 20, 60, 35, 55,46, 38],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
        "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})  
```

```python
df2.groupby("Department")["Salary"].mean()
df2.groupby("Team")["Age"].max()
```

```python
df2.groupby("Department")["Salary"].agg(
    avg_salary = "max",
    Age = "max",
    date = "count"
)
```

```python
# 18. Group data by a categorical column and calculate:
# 1. average
# 2. total

df2.groupby("Department")["Salary"].mean()
df2.groupby("Department")["Salary"].sum()
```

```python
# 19. Sort values by:
# 1. one column
# 2. multiple columns

df2.sort_values("Salary",ascending=False)
df2.sort_values(["Salary","Age"],ascending=False)
```

```python
#25. Merge two DataFrames:

# inner join
# left join
# right join

import pandas as pd
# DataFrame 1
df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Ram', 'Shyam', 'Mohan', 'Sita']
})

# DataFrame 2
df2 = pd.DataFrame({
    'ID': [3, 4, 5, 6],
    'Marks': [78, 88, 90, 95]
})

inner_join = pd.merge(df1,df2, on='ID',how="inner")
left_join = pd.merge(df1,df2, on='ID',how="left")   ## isse hum koi bhi particular column ko target kar ke concate kr skte h 
right_join = pd.merge(df1,df2, on='ID',how="right")


print(inner_join)
print("-------------------------------------")
print(left_join)
print("-------------------------------------")
print(right_join)
```

```python
# 26. Concatenate DataFrames:
# row-wise
# column-wise

row_ = pd.concat([df1,df2],axis=0)
column =  pd.concat([df1,df2],axis=1)   ## isme direct concat ho jayega without targeting any column
print(row_)
print(column)
```

```python
#27. Pivot and melt a DataFrame.
df3 = pd.DataFrame({
    "Name" : ["Satyajit sahu","Ayushi Kumari"],
    "Score" : [10,20],
    "Subject" : ["Science","Commerece"]
})
print(df3)
pivot = df3.pivot_table(index="Name", columns="Subject", values="Score", aggfunc="mean")
print(pivot)
melt = pivot.reset_index().melt(id_vars="Name", var_name="Subject", value_name="Score")
print(melt)
```

```python
#28. Given student marks:
# find topper
# find average
# count students above average
import pandas as pd
df = pd.read_csv("da.csv")
print(df)
```

```python
count = df["Name"].count()
marks = df["Marks"].sum()
average = marks / count
print(average)
```

```python
above_avg = 0
for i in range(len(df)):
    if df["Marks"][i] > average :
        above_avg+=1
print(above_avg)
        
```

```python
max_= 0 
for i in range(len(df)) :
    if max_ < df["Marks"][i] :
        max_ = df["Marks"][i]
        result = df.loc[i,["Name","Marks", "Subject"]]
print(result)
```
