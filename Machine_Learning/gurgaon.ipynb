```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import StratifiedShuffleSplit

df = pd.read_csv("housing.csv")

df["Income_Cat"] = pd.cut(df["median_income"],bins=[0, 1.5, 3.0, 4.5, 6.0, np.inf],labels=[1,2,3,4,5]).astype(int)


split = StratifiedShuffleSplit(n_splits=1,test_size=0.2,random_state=42)

for train_index, test_index in split.split(df,df["Income_Cat"]) :
    train_data = df.loc[train_index]
    test_data = df.loc[test_index]
    
```

```python
for sett in train_data,test_data :
    sett.drop(columns="Income_Cat",inplace = True,axis=1)
```

```python
df = train_data.copy()
df
```

```python
df.plot(kind = "scatter",x="latitude",y="longitude",alpha=0.2,grid=True,cmap ="jet",c="median_income") ##VVI
```

```python
## OneHotEncoder

from sklearn.preprocessing import OneHotEncoder

one_hot = OneHotEncoder()
Categorical = one_hot.fit_transform(df[["ocean_proximity"]])
housing_cat = pd.DataFrame(Categorical.toarray(), columns=['<1H OCEAN', 'INLAND', 'ISLAND', 'NEAR BAY', 'NEAR OCEAN'],index = df.index).copy()
df = pd.concat([df,housing_cat],axis=1)
df.drop(columns="ocean_proximity",axis=1,inplace=True)


## IMPUTER

from sklearn.impute import SimpleImputer

impute = SimpleImputer(strategy="mean")
# num_cols = df.select_dtypes(include=['int64','float64']).columns   ## Only numerical columns
impute_data = impute.fit_transform(df)
df = pd.DataFrame(impute_data, columns=df.columns, index=df.index)
df
```

```python
from sklearn.preprocessing import MinMaxScaler

min_max = MinMaxScaler(feature_range=(-1,1))
featured_data=  min_max.fit_transform(df)

df  = pd.DataFrame(featured_data,columns=df.columns,index=df.index)
```
