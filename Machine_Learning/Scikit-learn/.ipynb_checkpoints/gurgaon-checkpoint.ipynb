```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# ------------------------------
# Load & preprocess
# ------------------------------
df = pd.read_csv("gurugram_train.csv")
df.drop(columns="Unnamed: 9", inplace=True)

y = df['Price']
df['Rate per sqft'] = (
    df['Rate per sqft']
    .astype(str)
    .str.replace(',', '', regex=True)
    .str.strip()
    .astype(float)
)

# Numeric features
x_already_numeric = df[['Area','BHK_Count','Rate per sqft']]

# Categorical features
x_has_numeric = df[['Status', 'Property Type', 'Locality', 
            'Builder Name', 'RERA Approval', 'Socity', 
            'Company Name', 'Flat Type']]
x_encoded = pd.get_dummies(x_has_numeric, drop_first=True)

# Concatenate numeric + categorical
x = pd.concat([x_already_numeric, x_encoded], axis=1)

# ------------------------------
# Scale numeric features
# ------------------------------
scaler = StandardScaler()
x_already_numeric_scaled = scaler.fit_transform(x_already_numeric)
x_scaled = np.concatenate([x_already_numeric_scaled, x_encoded.values], axis=1)

# ------------------------------
# Train/test split
# ------------------------------
X_train, X_test, y_train, y_test = train_test_split(x_scaled, np.log1p(y), test_size=0.2, random_state=42)

# ------------------------------
# Train model
# ------------------------------
clf = LinearRegression()
clf.fit(X_train, y_train)
print("Model trained Successfully !!!")

# ------------------------------
# Evaluate model
# ------------------------------
y_pred_log = clf.predict(X_test)
y_pred = np.expm1(y_pred_log)

r2 = r2_score(y_test, y_pred_log)   # compare in log scale
mae = mean_absolute_error(np.expm1(y_test), y_pred)  # back to original scale
rmse = np.sqrt(mean_squared_error(np.expm1(y_test), y_pred))

print("R² Score:", r2)
print("MAE:", mae)
print("RMSE:", rmse)

# ------------------------------
# New data prediction
# ------------------------------
new_data = {
    'Area': [1528],
    'BHK_Count': [3],
    'Rate per sqft': [9450],
    'Status': ['Under Construction'],
    'Property Type': ['2 BHK Apartment in M3M Antalya Hills Phase I'],
    'Locality': ['Sector 79'],
    'Builder Name': ['Property In Gurgaon'],
    'RERA Approval': ['Approved by RERA'],
    'Socity': ['M3M Antalya Hills Phase I'],
    'Company Name': ['M3M'],
    'Flat Type': ['Apartment']
}

new_df = pd.DataFrame(new_data)
new_df_encoded = pd.get_dummies(new_df)

# Ensure same columns as training
new_df_encoded = new_df_encoded.reindex(columns=x_encoded.columns, fill_value=0)

# Scale numeric features
new_scaled_numeric = scaler.transform(new_df[['Area','BHK_Count','Rate per sqft']])
new_scaled = np.concatenate([new_scaled_numeric, new_df_encoded.values], axis=1)

# Predict
pred_log = clf.predict(new_scaled)
predicted_price = np.expm1(pred_log)
print("Predicted Price:", predicted_price)


```
