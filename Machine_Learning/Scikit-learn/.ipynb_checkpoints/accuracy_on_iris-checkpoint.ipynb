```python
import pandas as pd

df = pd.read_csv("iris_train.csv")
df
```

```python
x = df.iloc[:,:-1]
y = df.iloc[:,-1]


from sklearn.tree import DecisionTreeClassifier


clf = DecisionTreeClassifier()
clf.fit(x,y)

predictions = clf.predict([[4.6,3.1,1.5,0.2]])
predictions
```

```python
x_test = df.iloc[:,:-1]
prediction = clf.predict(x_test)
prediction
```

```python
y_test= df.iloc[:,-1]
y_test.values
```

```python
count = 0
for i in range(0,len(y_test)):
    if prediction[i] == y_test[i] :
        count = count+1

Accuracy = (count*100)/len(y_test)
print(Accuracy)
```
