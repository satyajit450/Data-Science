```python
from sklearn.tree import DecisionTreeClassifier  ## use to classify things ex:- cat / dog
```

```python
features = [
    [150, 0],  # Apple
    [170, 0],  # Apple
    [130, 1],  # Orange
    [120, 1],  # Orange
    [10, 2],  # Mango
    [60, 2],  # Mango

]

labels = ["apple","orange","apple","orange","Mango","Mango"]
clf = DecisionTreeClassifier()
clf = clf.fit(features,labels)
print(clf.predict([[190,1]]))
print(clf.predict([[180,0]]))
print(clf.predict([[160,1]]))
print(clf.predict([[50,0]]))
print(clf.predict([[40,2]]))
print(clf.predict([[199,0]]))
print(clf.predict([[10,2]]))

```
