```python
from sklearn.tree import DecisionTreeClassifier  ## use to classify things ex:- cat / dog
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

# | Situation           | Best Choice                 |
# | ------------------- | --------------------------- |
# | Small dataset       | Logistic / KNN              |
# | Text data           | Naive Bayes / SVM           |
# | Tabular data        | **XGBoost / Random Forest** |
# | Need explainability | Logistic / Decision Tree    |
# | High accuracy goal  | **Gradient Boosting**       |
# | Images / NLP        | Neural Networks             |
```

```python
features = [
    [150, 0],  # Apple
    [170, 0],  # Apple
    [130, 1],  # Orange
    [120, 1],  # Orange
    [10, 2],  # Mango
    [60, 2],  # Mango

]

labels = ["apple","orange","apple","orange","Mango","Mango"]
clf = DecisionTreeClassifier()
clf = clf.fit(features,labels)   #fit the data into the model
print(clf.predict([[190,1]]))
print(clf.predict([[180,0]]))
print(clf.predict([[160,1]]))
print(clf.predict([[50,0]]))
print(clf.predict([[40,2]]))
print(clf.predict([[199,0]]))
print(clf.predict([[10,2]]))
```

```python
plt.figure(figsize=(20,10))
plot_tree(clf,feature_names=["Weight","Names"],class_names=["Apple","Orange","Mango"],filled=True,rounded=True,fontsize=20)
```
