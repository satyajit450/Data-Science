```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [
    [150, 0],  # Apple
    [170, 0],  # Apple
    [130, 1],  # Orange
    [120, 1],  # Orange
]

labels = ["apple","orange","apple","orange"]
clf = DecisionTreeClassifier()
clf = clf.fit(features,labels)
print(clf.predict([[150,0]]))
```
