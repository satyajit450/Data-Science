```python
# 1.Create a line plot for
# x = [1,2,3,4,5]
# y = [2,4,1,8,7]
# Add:
# title
# x-label and y-label
# grid

import matplotlib.pyplot as plt 
x = [1,2,3,4,5]
y = [2,4,1,8,7]

plt.plot(x,y,linestyle="--",color="black",marker="^",markeredgecolor="tab:red",markerfacecolor='tab:green')
plt.title("LinePlot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
```

```python
# 2.Plot two line graphs in the same figure:
# y1 = x²
# y2 = x³
y1  = [i * i for i in x]
y2  = [i * i * i for i in x]
z2  = [i * i *i* i for i in x]

print(y1)
print(y2)
plt.plot(y1,marker="X")
plt.plot(y2,marker="o")
plt.plot(z2,color="red",marker="^")
plt.grid()
```

```python
import numpy as np
import matplotlib.pyplot as plt

sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]

width = 0.25
x = np.arange(len(years))
# print(x)

plt.bar(x + width, kohli, width=width, label="Kohli",color="#090531",edgecolor="black")
plt.bar(x, sehwag, width=width, label="Sehwag",color="darkgreen",edgecolor="black")
plt.bar(x - width, sachin, width=width, label="Sachin",color="red",edgecolor="black",alpha=1)

plt.xticks(x, years)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.legend()
plt.tight_layout()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

# Sales data for each product
product_a = [20, 25, 30, 28, 35, 40]
product_b = [15, 18, 20, 25, 30, 35]
product_c = [25, 28, 30, 35, 40, 45]


plt.stackplot(months,product_a,product_b,product_c, labels=['Product A', 'Product B', 'Product C'], colors=['#FF5733','#33FF57','#3357FF'])
plt.grid()
plt.legend(loc="best")
```

```python
import matplotlib.pyplot as plt

# Single figure with 1 row and 2 columns
plt.figure(figsize=(6,6))  # Width bada rakha taaki dono side-by-side clearly dikhe

# First pie chart
plt.subplot(1, 2, 1)     #subplot(row,column,plot_number)
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_a = [20, 25, 30, 28, 35, 40]
plt.pie(product_a, labels=months, shadow=True, autopct='%1d%%')
plt.title("Product A Monthly Distribution")

# Second pie chart
plt.subplot(1, 2, 2)
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
ac = [12, 14, 15, 13, 16, 18, 20]
plt.pie(ac, labels=days, shadow=True, autopct='%1d%%')
plt.title("AC Usage Weekly")

plt.tight_layout()  # Spacing adjust
plt.show()
```

```python
import matplotlib.pyplot as plt

fig,axes = plt.subplots(1,2,figsize=(12,6))
axes[0].plot(months,product_a,linestyle="--",color="green",marker="^")
axes[0].set_xlabel("Months")
axes[0].set_ylabel("Sells")

axes[1].plot(days,ac,linestyle="--",color="green",marker="^")
axes[1].set_xlabel("Days")
axes[1].set_ylabel("AC Sells")

plt.tight_layout()
fig.suptitle("SELLS", fontsize=16, y=1.05)
```
