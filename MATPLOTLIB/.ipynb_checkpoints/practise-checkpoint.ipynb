```python
#1.Create a line plot for
#x = [1,2,3,4,5]
#y = [2,4,1,8,7]
# Add:
# title
# x-label and y-label
# grid

import matplotlib.pyplot as plt 
x = [1,2,3,4,5]
y = [2,4,1,8,7]

line = plt.plot(x,y,linestyle="--",color="green",marker="^")
plt.title("LinePlot")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.grid()
```

```python
#2.Plot two line graphs in the same figure:
# y1 = x²
# y2 = x³
y1  = [i * i for i in x]
y2  = [i * i * i for i in x]
print(y1)
print(y2)
plt.plot(y1)
plt.plot(y2)
```

```python
import numpy as np
import matplotlib.pyplot as plt

sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]

width = 0.25
x = np.arange(len(years))
print(x)

plt.bar(x - width, sachin, width=width, label="Sachin",color="red",edgecolor="black",alpha=1)
plt.bar(x, sehwag, width=width, label="Sehwag",color="darkgreen",edgecolor="black")
plt.bar(x + width, kohli, width=width, label="Kohli",color="#090531",edgecolor="black")

plt.xticks(x, years)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.legend()
```

```python
#5.Create a histogram of 100 random numbers between 1 and 100.
import numpy as np

rand = np.random.randint(1,100,size=20)
np.random.seed(42)
max_ = np.max(rand)
min_ = np.min(rand)
print(rand)                                      # range = max - min
print(max_)
print(min_)
range_ = max_ - min_
bin_edges = [0, 17, 35, 53, 71, 88]
plt.hist(rand,bins=bin_edges,width=5,edgecolor='black')  #bins=5 ka matlab hai data ko 5 ranges me divide karo.
plt.axvline(20,color="red",linestyle="--")
plt.axhline(4,color="green",linestyle="--")
plt.figure(figsize=(10,10))
# plt.axline((5, 75), slope=0., color="green", linestyle="--")

```

```python
# Naming
import matplotlib.pyplot as plt

players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]

plt.bar(players, runs,width=0.50,edgecolor='black')
plt.xlabel("Players")
plt.ylabel("Runs")

for i in range(len("players")) :
    plt.text(i, runs[i] + 30, str(runs[i]), ha='center') #plt.text(x,y,str)

```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

# Sales data for each product
product_a = [20, 25, 30, 28, 35, 40]
product_b = [15, 18, 20, 25, 30, 35]
product_c = [25, 28, 30, 35, 40, 45]


plt.stackplot(months,product_a,product_b,product_c, labels=['Product A', 'Product B', 'Product C'], colors=['#FF5733','#33FF57','#3357FF'])
plt.grid()
plt.legend(loc="best")
```

```python
import matplotlib.pyplot as plt

# Days of the week
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

# Energy consumption data (kWh)
ac = [12, 14, 15, 13, 16, 18, 20]
fridge = [8, 8, 8, 8, 8, 8, 8]
lights = [5, 6, 5, 7, 6, 8, 7]

plt.stackplot(days,ac,fridge,lights,labels=["AC","Fridge","Lights"],colors=['#FF5733','#33FF57','#3357FF'],edgecolor="black")
plt.grid()
plt.legend()
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Sample data
data = pd.DataFrame({
    'Days': ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    'AC': [12, 14, 15, 13, 16, 18, 20],
    'Fridge': [8, 8, 8, 8, 8, 8, 8],
    'Lights': [5, 6, 5, 7, 6, 8, 7]
})

# Melt data for seaborn
data_melted = data.melt(id_vars='Days', var_name='Appliance', value_name='kWh')

# Line plot with Seaborn
sns.lineplot(data=data_melted, x='Days', y='kWh', hue='Appliance', marker='o')
plt.title('Daily Energy Consumption')
plt.show()
```
