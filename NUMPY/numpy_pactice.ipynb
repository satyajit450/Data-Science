```python
#22. Given a marks array of 50 students:
# find topper
# find average
# count students above average

np.random.seed(42)
arr = np.random.randint(1,50,size= 50)
print(arr)
topper = np.max(arr)
print("Topper is :",topper)

avg = np.average(arr)
print("Average is :",avg)

above_avg = arr[arr > avg ]
print("Above average :",above_avg)
```
