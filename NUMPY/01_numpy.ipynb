# NUMPY

```python
import numpy as np 
import time

size = 1_000
arr1 = np.array(range(size))

start = time.time()
square = arr1**2
end= time.time()

print ("Time"  , end-start)
# print(square)
```

# LIST

```python
list_1= list(range(size))
start = time.time()
square  = [sq**2 for sq in list_1]
end = time.time()
print ("Time"  , end-start)
```
